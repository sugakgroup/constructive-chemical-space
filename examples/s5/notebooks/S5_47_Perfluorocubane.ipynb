The nucleus indices below are *temporary display and construction addresses*, **not** intrinsic identifiers. All H nuclei and connections are displayed by default. The allocations are specified formal descriptions; these figures do not assert a uniquely preferred physical electronic state.


```python
from pathlib import Path
import sys

# Works when the notebook is launched from the repository root or notebooks/ directory.
cwd = Path.cwd().resolve()
repo_root = next((p for p in (cwd, *cwd.parents)
                  if (p / 'examples' / 's5' / '_common.py').is_file()), None)
if repo_root is None:
    raise FileNotFoundError('Open this notebook from the CCS repository tree.')
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))
from examples.s5._common import build_example, summarize_example, verify_example, export_example

example = build_example(47)
print(summarize_example(example))
replay_lengths = verify_example(example)
print('Exact W(D) replay verified:', replay_lengths)
```

## View constitution and electron allocations

The constitution uses direct-connection strokes, while the allocation view preserves endpoint electron contributions. RDKit is used only for non-normative depiction.


```python
# These options affect only drawing, not the CCS identity or electron allocations.
draw_options = dict(order=example.order, show_hydrogens=True,
                    show_indices=True, show_total_charge=True,
                    width=1500 if len(example.order) > 60 else 950,
                    height=1000 if len(example.order) > 60 else 650)
print('Constitution — direct-connection graph (every explicit H retained)')
example.constitution.show(**draw_options)
```

```python
for letter, form in example.forms.items():
    word = form.to_word(order=example.order)
    print(f'Form {letter}: {len(word.tokens)} core tokens, N_e = {form.num_electrons}')
    form.show(**draw_options)
```

Exports are disabled by default. SVG/PNG files are written under `examples/s5/outputs/` only if requested.


```python
# Set EXPORT = True to write reproducible SVG/PNG files outside version control.
EXPORT = False
if EXPORT:
    saved_files = export_example(example, formats=('svg',), **{
        key: value for key, value in draw_options.items() if key != 'order'
    })
    for path in saved_files:
        print(path)
```
